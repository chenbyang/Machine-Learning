```
from sklearn import datasets, svm                        # 匯入資料集與支援向量機模組
from sklearn.model_selection import train_test_split    # 匯入資料分割工具

# 載入鳶尾花（Iris）資料集
iris = datasets.load_iris()
x = iris.data        # 特徵資料（花萼長度、花萼寬度、花瓣長度、花瓣寬度）
y = iris.target      # 目標標籤（花的品種：0=setosa, 1=versicolor, 2=virginica）

# 分割資料集為訓練集與測試集，比例為 70% 訓練 / 30% 測試，設定亂數種子確保結果可重現
train_data, test_data, train_label, test_label = train_test_split(x, y, test_size=0.3, random_state=0)

# 建立支援向量機分類器（使用多項式核函數 'poly'）
# gamma='auto'：自動設定 gamma 參數為 1 / 特徵數
# C=100：錯誤懲罰參數（越大越不容忍錯誤，容易過擬合）
clf = svm.SVC(kernel='poly', gamma='auto', C=100)

# 使用訓練資料進行模型訓練
clf.fit(train_data, train_label)

# 輸出模型在訓練資料上的準確率
print("訓練資料正確率:", clf.score(train_data, train_label))

# 輸出模型在測試資料上的準確率
print("測試資料正確率:", clf.score(test_data, test_label))

# 使用模型對測試資料進行預測
predict = clf.predict(test_data)

# 輸出測試資料的真實標籤（標準答案）
print("標準答案:", test_label)

# 輸出模型預測的標籤（預測答案）
print("預測答案:", predict)
```
