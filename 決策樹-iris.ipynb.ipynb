```python
# 從sklearn匯入所需的模組
from sklearn import tree
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# 讀入鳶尾花資料集
iris = load_iris()  # 將鳶尾花資料集載入為一個字典物件
iris_data = iris.data  # 特徵資料 (花萼長度、花萼寬度、花瓣長度、花瓣寬度)
iris_label = iris.target  # 標籤資料 (對應的三種類別: Setosa, Versicolor, Virginica)

# 將資料切分成訓練資料與測試資料，這裡設置測試集佔 30%
train_data, test_data, train_label, test_label = train_test_split(iris_data, iris_label, test_size=0.3)

# 建立一個決策樹分類器
clf = tree.DecisionTreeClassifier()  # 使用預設參數建立一個決策樹分類器
iris_clf = clf.fit(train_data, train_label)  # 使用訓練資料進行模型訓練

# 計算訓練資料的預測正確率
print("訓練資料正確率:", clf.score(train_data, train_label))  # 訓練資料正確率

# 計算測試資料的預測正確率
print("測試資料正確率:", clf.score(test_data, test_label))  # 測試資料正確率

# 使用訓練好的分類器對測試資料進行預測
predict = clf.predict(test_data)

# 顯示測試資料的標準答案 (實際標籤)
print("標準答案:", test_label)

# 顯示模型預測的答案
print("預測答案:", predict)
```
