```python
# 從sklearn匯入隨機森林分類器、鳶尾花資料集與資料切分工具
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# 讀入鳶尾花資料集
iris = load_iris()  # 這會將鳶尾花資料集載入到一個字典物件中
iris_data = iris.data  # 特徵資料 (4個特徵: 花萼長度、花萼寬度、花瓣長度、花瓣寬度)
iris_label = iris.target  # 標籤資料 (對應的三個花的種類)

# 將資料切分成訓練資料與測試資料，這裡設置測試集佔 30%
train_data, test_data, train_label, test_label = train_test_split(iris_data, iris_label, test_size=0.3)

# 建立隨機森林分類器，這裡使用30顆樹，並設定每顆樹的最大深度為4
clf = RandomForestClassifier(n_estimators=30, max_depth=4)

# 使用訓練資料進行模型訓練
iris_clf = clf.fit(train_data, train_label)

# 訓練資料的預測正確率
print("訓練資料正確率:", clf.score(train_data, train_label))  # 訓練資料正確率

# 測試資料的預測正確率
print("測試資料正確率:", clf.score(test_data, test_label))  # 測試資料正確率

# 使用訓練好的分類器對測試資料進行預測
predict = clf.predict(test_data)

# 顯示測試資料的標準答案 (實際標籤)
print("標準答案:", test_label)

# 顯示模型預測的答案
print("預測答案:", predict)
```
